```python
import numpy as np
import matplotlib.pyplot as plt
# ---------------------------------------------------
# Input Signal
# ---------------------------------------------------
signal = np.array([1, 2, 3, 4, 5, 6, 7, 8])
n = np.arange(len(signal))
# ---------------------------------------------------
# FFT
# ---------------------------------------------------
fft_result = np.fft.fft(signal)
real_part = np.real(fft_result)
imaginary_part = np.imag(fft_result)
# Magnitude and Phase
magnitude_spectrum = np.abs(fft_result)
phase_spectrum = np.angle(fft_result)
# ---------------------------------------------------
# IFFT
# ---------------------------------------------------
reconstructed_signal = np.real(np.fft.ifft(fft_result))

# ===================================================
# CREATE FIGURE
# ===================================================
plt.figure(figsize=(14, 15))

# ===================================================
# 1. ORIGINAL SIGNAL
# ===================================================
plt.subplot(3, 2, 1)
plt.stem(n, signal)
plt.scatter(n, signal, s=45, zorder=3)
# Add data-point values
for x, y in zip(n, signal):
    plt.annotate(
        f'{y}',
        (x, y),
        textcoords="offset points",
        xytext=(0, 8),
        ha='center'
    )
plt.title("Original Signal", fontsize=14)
plt.xlabel("Sample Index (n)")
plt.ylabel("Amplitude")
plt.xticks(n)
plt.grid(True, alpha=0.3)

# ===================================================
# 2. REAL + IMAGINARY PART OF FFT
# ===================================================
plt.subplot(3, 2, 2)
plt.stem(
n, real_part,
linefmt='C0-',
markerfmt='C0o',
basefmt='C0-',
label="Real Part"
)
plt.stem(
n, imaginary_part,
linefmt='C1-',
markerfmt='C1o',
basefmt='C1-',
label="Imaginary Part"
)
# Add Real FFT values
for x, y in zip(n, real_part):
    plt.annotate(
f'{y:.2f}',
(x, y),
textcoords="offset points",
xytext=(0, 8),
ha='center',
fontsize=8
)
# Add Imaginary FFT values
for x, y in zip(n, imaginary_part):
    plt.annotate(
        f'{y:.2f}',
        (x, y),
        textcoords="offset points",
        xytext=(0, -15),
        ha='center',
        fontsize=8
    )
plt.title("Real and Imaginary Parts of FFT", fontsize=14)
plt.xlabel("Frequency Index (k)")
plt.ylabel("Amplitude")
plt.xticks(n)
plt.legend()
plt.grid(True, alpha=0.3)

# ===================================================
# 3. MAGNITUDE SPECTRUM
# ===================================================
plt.subplot(3, 2, 3)
plt.stem(n, magnitude_spectrum)
plt.scatter(n, magnitude_spectrum, s=45, zorder=3)
# Add magnitude values
for x, y in zip(n, magnitude_spectrum):
    plt.annotate(
        f'{y:.2f}',
        (x, y),
        textcoords="offset points",
        xytext=(0, 8),
        ha='center',
        fontsize=8
    )
plt.title("Magnitude Spectrum", fontsize=14)
plt.xlabel("Frequency Index (k)")
plt.ylabel("Magnitude")
plt.xticks(n)
plt.grid(True, alpha=0.3)

# ===================================================
# 4. PHASE SPECTRUM
# ===================================================
plt.subplot(3, 2, 4)
plt.stem(n, phase_spectrum)
plt.scatter(n, phase_spectrum, s=45, zorder=3)
# Add phase values
for x, y in zip(n, phase_spectrum):
    plt.annotate(
        f'{y:.2f}',
        (x, y),
        textcoords="offset points",
        xytext=(0, 8),
        ha='center',
        fontsize=8
    )
plt.title("Phase Spectrum", fontsize=14)
plt.xlabel("Frequency Index (k)")
plt.ylabel("Phase (radians)")
plt.xticks(n)
plt.grid(True, alpha=0.3)

# ===================================================
# 5. RECONSTRUCTED SIGNAL
# ===================================================
plt.subplot(3, 2, 5)
plt.stem(n, reconstructed_signal)
plt.scatter(n, reconstructed_signal, s=45, zorder=3)
# Add reconstructed values
for x, y in zip(n, reconstructed_signal):
    plt.annotate(
        f'{y:.2f}',
        (x, y),
        textcoords="offset points",
        xytext=(0, 8),
        ha='center',
        fontsize=8
    )
plt.title("Reconstructed Signal using IFFT", fontsize=14)
plt.xlabel("Sample Index (n)")
plt.ylabel("Amplitude")
plt.xticks(n)
plt.grid(True, alpha=0.3)

# ===================================================
# ADJUST SPACING
# ===================================================
plt.subplots_adjust(
hspace=0.50,
wspace=0.30,
top=0.95,
bottom=0.05,
left=0.08,
right=0.95
)
plt.show()

# ===================================================
# PRINT RESULTS
# ===================================================
print("Original Signal:")
print(signal)
print("\nFFT Result:")
print(fft_result)
print("\nReal Part:")
print(real_part)
print("\nImaginary Part:")
print(imaginary_part)
print("\nMagnitude Spectrum:")
print(magnitude_spectrum)
print("\nPhase Spectrum:")
print(phase_spectrum)
print("\nReconstructed Signal:")
print(reconstructed_signal)
```
