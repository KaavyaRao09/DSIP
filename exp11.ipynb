```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab.patches import cv2_imshow
# Load the image
image = cv2.imread("morph.png", 0) # Load as grayscale
```

```python
# Define the kernel (structuring element)
kernel = np.ones((5, 5), np.uint8)

rect_kernel = cv2.getStructuringElement(cv2.MORPH_RECT, (5, 5))
ellipse_kernel = cv2.getStructuringElement(cv2.MORPH_ELLIPSE, (5, 5))
cross_kernel = cv2.getStructuringElement(cv2.MORPH_CROSS, (5, 5))
```

```python
# Perform Erosion
erosion = cv2.erode(image, kernel, iterations=1)
# Perform Dilation
dilation = cv2.dilate(image, kernel, iterations=1)
# Perform Opening (Erosion followed by Dilation)
opening = cv2.morphologyEx(image, cv2.MORPH_OPEN, kernel, iterations=4)
# Perform Closing (Dilation followed by Erosion)
closing = cv2.morphologyEx(image, cv2.MORPH_CLOSE, kernel)
```

```python

# Display the original image and the results vertically

plt.figure(figsize=(6, 15))

plt.subplot(5, 1, 1)
plt.title('Original Image')
plt.imshow(image, cmap='gray')
plt.axis('off')

plt.subplot(5, 1, 2)
plt.title('Erosion')
plt.imshow(erosion, cmap='gray')
plt.axis('off')

plt.subplot(5, 1, 3)
plt.title('Dilation')
plt.imshow(dilation, cmap='gray')
plt.axis('off')

plt.subplot(5, 1, 4)
plt.title('Opening')
plt.imshow(opening, cmap='gray')
plt.axis('off')

plt.subplot(5, 1, 5)
plt.title('Closing')
plt.imshow(closing, cmap='gray')
plt.axis('off')

plt.tight_layout()
plt.show()
```
